Aggregation
Aggregation reduces an array to a single summary value (sum, mean, etc.). NumPy's own functions (np.sum, np.mean, ...) are far faster than Python's built-ins on NumPy arrays — the timing demo below shows the difference.

```python
a1
```

```python
sum(a1)
```

```python
# NumPy's sum() - same result, but optimised for arrays
np.sum(a1)
```

```python
# Create a large array to compare performance
massive_array = np.random.random(1000000)
massive_array.size, type(massive_array)
```

```python
# Time both: on a NumPy array, np.sum() is dramatically faster than Python's sum()
%timeit sum(massive_array)     # Python sum
%timeit np.sum(massive_array)  # NumPy sum
```

```python
# For contrast, build a plain Python list of the same size
import random
massive_list = [random.randint(0, 10) for i in range(1000000)]
len(massive_list), type(massive_list)
```

```python
# On a Python list the situation reverses: Python's sum() wins, because np.sum()
# must first convert the list to an array. Use NumPy functions on NumPy arrays.
%timeit sum(massive_list)     # Python sum
%timeit np.sum(massive_list)  # NumPy sum
```

Dot product
The main two rules for dot product to remember are:

The inner dimensions must match:
(3, 2) @ (3, 2) won't work
(2, 3) @ (3, 2) will work
(3, 2) @ (2, 3) will work
The resulting matrix has the shape of the outer dimensions:
(2, 3) @ (3, 2) -> (2, 2)
(3, 2) @ (2, 3) -> (3, 3)
Note: In NumPy, np.dot() and @ can be used to acheive the same result for 1-2 dimension arrays. However, their behaviour begins to differ at arrays with 3+ dimensions.

```python
# Create two matrices whose inner dimensions match: (3,3) and (3,2)
np.random.seed(0)
mat1 = np.random.randint(10, size=(3, 3))
mat2 = np.random.randint(10, size=(3, 2))

mat1.shape, mat2.shape
```

```python
mat1
```

```python
mat2
```

```python
np.dot(mat1, mat2)
```

```python
mat1 @ mat2
```
